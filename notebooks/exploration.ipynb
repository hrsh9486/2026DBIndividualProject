Initial exploration of yFinance library to gain understanding of India's key indices and performance.

```python
"""
Exploring how to use yfinance library
"""
import yfinance as yf
```

```python
# --- Top Indian companies by market cap vs global peers ---
top_india = ["RELIANCE.NS", "TCS.NS", "HDFCBANK.NS", "INFY.NS", "ICICIBANK.NS"]
global_peers = ["AAPL", "MSFT", "TSM", "JPM"]
```

```python
# --- Indices: India vs global peers ---
indices = {
    "Nifty50": "^NSEI",
    "Sensex": "^BSESN",
    "S&P500": "^GSPC",
    "HangSeng_China": "^HSI",
    "Bovespa_Brazil": "^BVSP",
    "EM_ETF": "EEM",       # broad emerging markets basket
}
```

```python
# --- Exchange Rates vs major economies ---
rates = {
    "Dollar to Rupee" : "USDINR=X",
    "Pound to Rupee" : "GBPINR=X",
    "Euro to Rupee" : "EURINR=X",
    "Yen to Rupee" : "JPYINR=X",
    "Franc to Rupee" : "CHFINR=X",
}
```

```python
print("=== 30-Year Index / FX Performance ===")
for name, ticker in indices.items():
    # Get 30 year price history for ticker 
    hist = yf.Ticker(ticker).history(period="30y", interval="1mo")
    if hist.empty:
        print(f"{name}: no data")
        continue

    # Calculate 30 year growth
    start, end = hist["Close"].iloc[0], hist["Close"].iloc[-1]
    total_return = (end / start - 1) * 100
    
    
    # If ticker is in currency other than USD, convert return
    currency = yf.Ticker(ticker).info["currency"]
    if currency != "USD":
        exchange_symbol = "USD" + currency + "=X"
        exchange_ticker = yf.Ticker(exchange_symbol)
        rate = exchange_ticker.history(period="1d")['Close'].iloc[-1]
        end = end / rate
    
    print(f"{name:18s} 30y return: {total_return:6.1f}%   latest: {end:,.2f} USD")
```

```python
print("\n=== Market Cap Snapshot (USD) ===")
for tk in top_india + global_peers:
    info = yf.Ticker(tk).info
    mcap = info.get("marketCap")
    currency = info.get("currency", "")
    name = info.get("shortName", tk)
    # If ticker is in currency other than USD, convert return
    if currency != "USD":
        exchange_symbol = "USD" + currency + "=X"
        exchange_ticker = yf.Ticker(exchange_symbol)
        rate = exchange_ticker.history(period="1d")['Close'].iloc[-1]
        umcap = mcap / rate
    else:
        exchange_symbol = "USDINR=X"
        exchange_ticker = yf.Ticker(exchange_symbol)
        rate = exchange_ticker.history(period="1d")['Close'].iloc[-1]
        umcap = mcap 
        mcap = mcap * rate
    
    
    if mcap:
        print(f"{name:35s} {tk:14s} {mcap/1e9:10,.1f}B INR  == {umcap/1e9:10,.1f}B USD")
```

```python
for name, ticker  in rates.items():
    hist = yf.Ticker(ticker).history(period="30y", interval="1mo")
    if hist.empty:
        print(f"{name}: no data")
        continue
    else:
        print(f"{name}")
        print(hist)
```
